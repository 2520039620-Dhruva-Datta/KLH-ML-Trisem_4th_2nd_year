```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)
```

```python
#Read the dataset
df = pd.read_csv(r"C:\Users\dhruv\OneDrive\KLH Univeristy\4th semester (Trimester Format KLH)\25SC2107E Machine Learning\Placement Prediction System (OS-SEC-2(2026-27))\data\titanic.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

## Select X and y from dataset

```python
X = df[["Pclass"]]
y = df["Fare"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## Use passengers who paid a fare only

Passengers with Fare = 0 (free tickets / crew) are removed, so only real fares are modelled

```python
filtered_df = df[df["Fare"] > 0].copy()

print("Rows with Fare > 0:", len(filtered_df))
```

```python
X = filtered_df[["Pclass"]]
y = filtered_df["Fare"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## Split the dataset into training and testing sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))
```

```python
model = LinearRegression()

#Traing the model with training data
model.fit(X_train, y_train)
```

```python
print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])
```

```python
#Predicting fare for training data

y_train_pred = model.predict(X_train)
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    X_train["Pclass"],
    y_train,
    alpha=0.5,
    label="Actual data"
)

plt.plot(
    X_train["Pclass"],
    y_train_pred,
    color="red",
    linewidth=3,
    label="Linear Regression Line"
)

plt.xlabel("Passenger Class (Pclass)")
plt.ylabel("Fare")
plt.title("Simple Linear Regression: Pclass → Fare")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Here the red line goes downward from left to right.

That means:

1. As Pclass increases (1st → 3rd class), the model predicts a lower fare on average.

2. So there is a negative relationship between Pclass and predicted fare (1st class is the most expensive).

The blue dots are the actual fares of passengers (they form three vertical columns because Pclass only takes the values 1, 2 and 3). The red line is the fare predicted using only Pclass. 

The downward slope indicates that lower class numbers pay higher fares. However, the large spread of blue points around the line (especially in 1st class) shows that Pclass alone cannot completely explain fare variation.

  This is why we later move from Simple Linear Regression to Multiple Linear Regression by adding other relevant features

```python
#predicting for testing data
y_test_pred = model.predict(X_test)
```

```python
mse = mean_squared_error(y_test, y_test_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_test_pred)
r2 = r2_score(y_test, y_test_pred)

print("MSE :", mse)
print("RMSE:", rmse)
print("MAE :", mae)
print("R²  :", r2)
```

```python
## Give a new passenger's Pclass (change the value below)
new_pclass = 1

## Predict Fare

predicted_value = model.predict(pd.DataFrame({"Pclass": [new_pclass]}))[0]

print(f"Predicted Fare: {predicted_value:.2f} fare units")
```
